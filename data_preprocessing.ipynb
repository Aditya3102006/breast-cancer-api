# Breast Cancer Classification & Model Training
This notebook loads the local `data.csv` file, preprocesses features to match the API and Streamlit UI configurations, trains a `RandomForestClassifier`, and outputs the pre-trained `model.pkl` and `features.pkl` files.

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
import pickle
```

```python
# Loading the data from local csv file
data_frame = pd.read_csv("data.csv")
```

```python
# Print the first 5 rows of the dataframe
data_frame.head()
```

```python
# Number of rows and columns in the dataset
data_frame.shape
```

```python
# Getting some information about the data
data_frame.info()
```

```python
# Checking for missing values
data_frame.isnull().sum()
```

```python
# Checking the distribution of Target Variable
data_frame['diagnosis'].value_counts()
```

```python
# Encode target variable: M = 1 (Malignant), B = 0 (Benign)
# This maps correctly to the Streamlit UI and FastAPI backend definitions
data_frame['label'] = data_frame['diagnosis'].map({'M': 1, 'B': 0})
data_frame.groupby('label').mean(numeric_only=True)
```

```python
# Separating the features and target
X = data_frame.drop(columns=['id', 'diagnosis', 'label'])
Y = data_frame['label']
```

```python
print(Y)
```

```python
# Splitting the data into training data & Testing data
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=2)
print(X.shape, X_train.shape, X_test.shape)
```

```python
# Model Training - RandomForestClassifier
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, Y_train)
```

```python
# Accuracy on training data
X_train_prediction = model.predict(X_train)
training_data_accuracy = accuracy_score(Y_train, X_train_prediction)
print('Accuracy on training data = ', training_data_accuracy)
```

```python
# Accuracy on test data
X_test_prediction = model.predict(X_test)
test_data_accuracy = accuracy_score(Y_test, X_test_prediction)
print('Accuracy on test data = ', test_data_accuracy)
```

```python
# Save the trained model and features to pkl files for Streamlit app and FastAPI
with open('model.pkl', 'wb') as f:
    pickle.dump(model, f)

with open('features.pkl', 'wb') as f:
    pickle.dump(list(X.columns), f)

print("Successfully saved model.pkl and features.pkl!")
```
